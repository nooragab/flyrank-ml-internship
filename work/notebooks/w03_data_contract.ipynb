## 1. Unit of analysis + time window

## Unit of Analysis + Time Window

**One row = one content item (`content_hash_id`)**, aggregated from its daily performance
records within a single mid-panel month, `month=2026-03`. The raw fact table's own grain is
finer — one row per (content_hash_id, report_date) per day — so I aggregate that daily grain
up to one row per page for this exercise, matching my lane's real decision grain: "should
this page be reviewed?" is a per-page question, not a per-day question.

**Tables used:**
- `dim_content.parquet` — content metadata (word_count, search_volume, content_type, dates)
- `fact_content_daily_performance/month=2026-03/` — daily GSC/GA4 facts for that month
- `dim_clients.parquet` — used only to check data availability, not as a feature source

**Time window:** March 2026 (`month=2026-03`), a single mid-panel month, per the assignment's
guidance to iterate on a middle month and treat the final month as sealed. Note: this is a
31-day window, shorter than the 90-day trailing window the starter dataset used.

**Target/proxy:** a simple within-month proxy — whether a page's GSC impressions fell from
the first half of March to the second half (`is_declining_month`). This is a placeholder
proxy for this contract exercise; a real capstone label would compare a full prior window to
a full future window, not two halves of one month.

**Deliberately excluded:** the warehouse ships no product decision flags at all (health_score,
priority_score) — there's nothing to strip there. What I exclude by choice is using
`content_updated_date` as a raw date feature; I convert it to `days_since_last_update` instead,
since a raw date isn't directly usable and invites sloppy handling.

```python
rel = "hf://datasets/FlyRank/internship-warehouse"

window_check = con.sql(f"""
    SELECT MIN(report_date) AS min_date, MAX(report_date) AS max_date, COUNT(*) AS n_rows
    FROM read_parquet('{rel}/fact_content_daily_performance/month=2026-03/*.parquet')
""").df()
window_check
```

## 3. Verify it with queries (grain, counts, missing values, windows)

*Every claim above gets a query cell here. A contract claim without a query next to it is a guess.*

### Three verification queries, five features, and the leakage trap

Every claim from Sections 1–2 gets a query here: the grain, the slice's size and date span,
and availability filtered with `IS TRUE`. Then a five-feature frame, and the deliberate
leakage experiment.

```python
grain_check = con.sql(f"""
    SELECT COUNT(*) AS total_rows,
           COUNT(DISTINCT (content_hash_id, report_date)) AS distinct_content_day_pairs
    FROM read_parquet('{rel}/fact_content_daily_performance/month=2026-03/*.parquet')
""").df()
print("If total_rows == distinct_content_day_pairs, grain confirmed: one row = one page-day.")
grain_check
```

```python
slice_check = con.sql(f"""
    SELECT COUNT(*) AS n_rows,
           COUNT(DISTINCT content_hash_id) AS n_unique_pages,
           MIN(report_date) AS min_date,
           MAX(report_date) AS max_date
    FROM read_parquet('{rel}/fact_content_daily_performance/month=2026-03/*.parquet')
    WHERE gsc_impressions > 0
""").df()
slice_check
```

```python
availability_check = con.sql(f"""
    SELECT COUNT(*) AS total_rows,
           COUNT(*) FILTER (WHERE gsc_data_available IS TRUE) AS gsc_available_rows,
           COUNT(*) FILTER (WHERE ga4_data_available IS TRUE) AS ga4_available_rows
    FROM read_parquet('{rel}/fact_content_daily_performance/month=2026-03/*.parquet')
""").df()
availability_check
```

```python
feature_frame = con.sql(f"""
    WITH monthly AS (
        SELECT
            content_hash_id,
            client_hash_id,
            SUM(gsc_impressions) AS impressions_month,
            SUM(gsc_clicks) AS clicks_month,
            AVG(gsc_avg_position) AS avg_position_month,
            SUM(CASE WHEN report_date < DATE '2026-03-16' THEN gsc_impressions ELSE 0 END) AS impressions_first_half,
            SUM(CASE WHEN report_date >= DATE '2026-03-16' THEN gsc_impressions ELSE 0 END) AS impressions_second_half
        FROM read_parquet('{rel}/fact_content_daily_performance/month=2026-03/*.parquet')
        WHERE gsc_data_available IS TRUE
        GROUP BY content_hash_id, client_hash_id
    )
    SELECT
        m.content_hash_id, m.client_hash_id,
        m.impressions_month, m.clicks_month, m.avg_position_month,
        ROUND(m.clicks_month / NULLIF(m.impressions_month, 0), 4) AS ctr_month,
        d.word_count, d.search_volume,
        DATE_DIFF('day', d.content_updated_date, DATE '2026-03-31') AS days_since_last_update,
        m.impressions_first_half, m.impressions_second_half,
        (m.impressions_second_half < m.impressions_first_half) AS is_declining_month
    FROM monthly m
    JOIN read_parquet('{rel}/dim_content.parquet') d ON m.content_hash_id = d.content_hash_id
    WHERE m.impressions_month > 0
""").df()

print(f"Feature frame shape: {feature_frame.shape}")
feature_frame.head(5)
```

### Five features, each "knowable at the decision moment because…"

1. **impressions_month** — observed GSC data throughout the window; known by month's end.
2. **avg_position_month** — an observed ranking signal, not a future outcome.
3. **ctr_month** — derived purely from impressions/clicks already observed.
4. **word_count** — static content metadata, fixed when the page was written.
5. **days_since_last_update** — computed from `content_updated_date`, known the moment you look at the page.

```python
# THE TRAP: add a column built from the same numbers the label came from
feature_frame["impressions_second_half_LEAKY"] = feature_frame["impressions_second_half"]

leaky_pred = feature_frame["impressions_second_half_LEAKY"] < feature_frame["impressions_first_half"]
quick_accuracy_leaky = (leaky_pred == feature_frame["is_declining_month"]).mean()
print(f"Quick 'accuracy' using the leaky feature alone: {quick_accuracy_leaky:.3f}  <- suspiciously perfect")

# Delete it and keep the honest frame
feature_frame = feature_frame.drop(columns=["impressions_second_half_LEAKY",
                                             "impressions_first_half", "impressions_second_half"])
print("Leaky columns removed. Honest feature frame:", list(feature_frame.columns))
```

**What happened:** `impressions_second_half` is literally the number my label was built
from — feeding it back in as a "feature" makes the quick score jump to (or near) perfect,
because the model isn't finding a pattern, it's just reading the answer. That's the same
leakage trap from notebook 02, reproduced here on real warehouse data. I removed the column
and kept only the five honest, pre-decision features.

## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*

## Data Limits

**Named limitation of this slice:** March 2026 is only 31 days, not a full 90-day trailing
window like the starter dataset used — some "decline" signals here may be noise that would
smooth out over a longer window.

**Unbalanced history:** clients' GSC/GA4 tracking start dates differ widely — some since
early 2025, others only mid-2026. A page from a client whose tracking started after March
would show `gsc_data_available = FALSE` for the whole month — not because it had no traffic,
but because there's no tracking data yet.

**GSC-only early rows:** rows before a client's `ga4_data_start` have `ga4_data_available = FALSE`
even with GSC data present, so GA4 features aren't usable for every row.

**What this data can never tell me:** why a page's impressions moved (algorithm change,
seasonality, a sibling page absorbing demand) — only observed direction and magnitude. And a
within-month "decline" here is a proxy for this exercise, not a genuine multi-month decline.

```python
unbalanced_check = con.sql(f"""
    SELECT
        COUNT(*) AS n_clients,
        COUNT(*) FILTER (WHERE gsc_data_start IS NULL) AS no_gsc_start,
        MIN(gsc_data_start) AS earliest_gsc_start,
        MAX(gsc_data_start) AS latest_gsc_start,
        MIN(ga4_data_start) AS earliest_ga4_start,
        MAX(ga4_data_start) AS latest_ga4_start
    FROM read_parquet('{rel}/dim_clients.parquet')
""").df()
unbalanced_check
```
